## 4. Missing Value Analysis

```python
# Count of missing values per column
df.isnull().sum()
```

```python
# Percentage of missing values per column
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage.sort_values(ascending=False)
```

## 5. Data Cleaning

We always create a copy of the original data before making major changes.

```python
# Create a copy of the original dataset
df_clean = df.copy()
df_clean.head()
```

### 5.1 Clean `Age`

Valid ages are considered between 18 and 100 for this project.

```python
# Inspect invalid ages
df_clean[(df_clean["Age"] < 0) | (df_clean["Age"] > 100)]
```

```python
# Convert invalid ages to missing values
df_clean.loc[(df_clean["Age"] < 18) | (df_clean["Age"] > 100), "Age"] = np.nan
```

```python
# Fill missing ages with the median
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())
df_clean["Age"].describe()
```

### 5.2 Clean `Quantity`

```python
# Inspect invalid quantities (should be greater than 0)
df_clean[df_clean["Quantity"] <= 0]
```

```python
# Convert invalid quantities to missing values
df_clean.loc[df_clean["Quantity"] <= 0, "Quantity"] = np.nan
```

```python
# Fill missing quantities with the median
df_clean["Quantity"] = df_clean["Quantity"].fillna(df_clean["Quantity"].median())
```

### 5.3 Clean `Discount_Percent`

```python
# Inspect invalid discount percentages (should be between 0 and 100)
df_clean[(df_clean["Discount_Percent"] < 0) | (df_clean["Discount_Percent"] > 100)]
```

```python
# Convert invalid discount percentages to missing values
df_clean.loc[(df_clean["Discount_Percent"] < 0) | (df_clean["Discount_Percent"] > 100), "Discount_Percent"] = np.nan
```

```python
# Fill missing discount percentages with the median
df_clean["Discount_Percent"] = df_clean["Discount_Percent"].fillna(df_clean["Discount_Percent"].median())
```

### 5.4 Clean `Customer_Rating`

Normally customer ratings should be between 1 and 5.

```python
# Inspect invalid ratings
df_clean[(df_clean["Customer_Rating"] < 1) | (df_clean["Customer_Rating"] > 5)]
```

```python
# Convert invalid ratings to missing values
df_clean.loc[(df_clean["Customer_Rating"] < 1) | (df_clean["Customer_Rating"] > 5), "Customer_Rating"] = np.nan
```

```python
# Fill missing ratings with the median
df_clean["Customer_Rating"] = df_clean["Customer_Rating"].fillna(df_clean["Customer_Rating"].median())
```

```python
# Confirm no invalid ratings remain
df_clean[(df_clean["Customer_Rating"] < 1) | (df_clean["Customer_Rating"] > 5)]
```

### 5.5 Fill Missing Categorical Values

Missing categorical values are filled with `"Unknown"`.

```python
# Check categorical columns
categorical_columns = df_clean.select_dtypes(include="object").columns
categorical_columns
```

```python
# Fill missing categorical values with "Unknown"
categorical_missing_columns = [
    "Customer_Name",
    "Gender",
    "Customer_Segment",
    "State",
    "City",
    "Sub_Category",
    "Payment_Method"
]

for column in categorical_missing_columns:
    df_clean[column] = df_clean[column].fillna("Unknown")
```

```python
# Confirm missing values are handled
df_clean.isnull().sum()
```

### 5.6 Convert Date Columns

```python
# Convert Order_Date to datetime (invalid dates become NaT)
df_clean["Order_Date"] = pd.to_datetime(df_clean["Order_Date"], errors="coerce")
df_clean["Order_Date"].isna().sum()
```

```python
# Drop rows where Order_Date could not be parsed
df_clean = df_clean.dropna(subset=["Order_Date"])
df_clean["Order_Date"].isnull().sum()
```

```python
# Convert Delivery_Date to datetime
df_clean["Delivery_Date"] = pd.to_datetime(df_clean["Delivery_Date"])
df_clean.dtypes
```

### 5.7 Create `Delivery_Days`

```python
# Number of days between order and delivery
df_clean["Delivery_Days"] = (df_clean["Delivery_Date"] - df_clean["Order_Date"]).dt.days
df_clean[["Order_Date", "Delivery_Date", "Delivery_Days"]].head()
```

## 6. Feature Engineering

```python
# Extract order year and month
df_clean["Order_Year"] = df_clean["Order_Date"].dt.year
df_clean["Month_Number"] = df_clean["Order_Date"].dt.month
```

**Profit Margin** = Profit / Sales × 100

```python
# Calculate profit margin
df_clean["Profit_Margin"] = (df_clean["Profit"] / df_clean["Sales"]) * 100
```

```python
# Create age groups
df_clean["Age_Group"] = pd.cut(
    df_clean["Age"],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "65+"]
)
df_clean["Age_Group"].value_counts()
```

## 7. Key Performance Indicators (KPIs)

```python
# Calculate overall business KPIs
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order_ID"].nunique()
total_customers = df_clean["Customer_ID"].nunique()
avg_rating = df_clean["Customer_Rating"].mean()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Average Rating:", avg_rating)
```

## 9. Final Dashboard

```python
fig, axes = plt.subplots(3, 2, figsize=(20, 18))

# -----------------------------------
# 1. SALES BY CATEGORY
# -----------------------------------
category_sales = (
    df_clean.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
sns.barplot(x=category_sales.values, y=category_sales.index, ax=axes[0, 0])
axes[0, 0].set_title("Sales by Category")

# -----------------------------------
# 2. PROFIT BY CATEGORY
# -----------------------------------
category_profit = (
    df_clean.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
sns.barplot(x=category_profit.values, y=category_profit.index, ax=axes[0, 1])
axes[0, 1].set_title("Profit by Category")

# -----------------------------------
# 3. MONTHLY SALES TREND
# -----------------------------------
monthly_sales = (
    df_clean.groupby(df_clean["Order_Date"].dt.to_period("M"))["Sales"]
    .sum()
    .sort_index()
)
monthly_sales.index = monthly_sales.index.astype(str)

axes[1, 0].plot(monthly_sales.index, monthly_sales.values)
axes[1, 0].set_title("Monthly Sales Trend")
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Sales")
axes[1, 0].tick_params(axis="x", rotation=45)

# -----------------------------------
# 4. TOP 10 PRODUCTS
# -----------------------------------
top_products = (
    df_clean.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
sns.barplot(x=top_products.values, y=top_products.index, ax=axes[1, 1])
axes[1, 1].set_title("Top 10 Products by Sales")

# -----------------------------------
# 5. SALES BY CUSTOMER SEGMENT
# -----------------------------------
segment_sales = df_clean.groupby("Customer_Segment")["Sales"].sum()
sns.barplot(x=segment_sales.index, y=segment_sales.values, ax=axes[2, 0])
axes[2, 0].set_title("Sales by Customer Segment")

# -----------------------------------
# 6. ORDER STATUS
# -----------------------------------
sns.countplot(data=df_clean, x="Order_Status", ax=axes[2, 1])
axes[2, 1].set_title("Order Status Distribution")

# -----------------------------------
# MAIN TITLE
# -----------------------------------
fig.suptitle("E-Commerce Sales & Profit Dashboard", fontsize=22, fontweight="bold")

plt.tight_layout()
plt.show()
```
